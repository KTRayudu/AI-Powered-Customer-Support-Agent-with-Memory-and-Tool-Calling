```python
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
from mem0 import Memory

mem0_config = {
    "vector_store": {
        "provider": "qdrant",
        "config": {"path":"./data/mem0-qdrant",
                   "embedding_model_dims":384},
    },
    "llm": {
        "provider": "groq",
        "config": {"model": "llama-3.3-70b-versatile","api_key": os.environ["GROQ_API_KEY"], "temperature": 0.1},
    },
    "embedder": {
        "provider": "huggingface",
        "config": {"model": "all-MiniLM-L6-v2"},
    },
}

memory = Memory.from_config(mem0_config)
```

```python
messages = [
    {"role": "user", "content": "Hi, I'm Alex. I love basketball and gaming."},
    {"role": "assistant", "content": "Hey Alex! I'll remember your interests."}
]
memory.add(messages, user_id="alex")
```

```python
results = memory.search("What do you know about me?", user_id="alex")
print(results)
```
